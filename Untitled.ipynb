```python
db_path = PROJECT_ROOT / "data/runtime/trailsight.duckdb"

con = duckdb.connect(str(db_path), read_only=True)

print("Connected to:")
print(db_path)
```

```python
con.execute("SHOW TABLES").df()
```

```python
for table in ["cases", "transactions", "case_transactions", "entities"]:
    print(f"\n{'=' * 60}")
    print(table.upper())
    print("=" * 60)
    
    display(con.execute(f"DESCRIBE {table}").df())
```

```python
cases = con.execute("""
    SELECT *
    FROM cases
""").df()

cases
```

```python
transactions = con.execute("""
    SELECT *
    FROM transactions
    LIMIT 20
""").df()

transactions
```

```python
columns = con.execute("""
    DESCRIBE transactions
""").df()["column_name"].tolist()

print(columns)

print("\nIs Laundering present?")
print("Is Laundering" in columns)
```

```python
demo_case_transactions = con.execute("""
    SELECT *
    FROM case_transactions
    WHERE case_ref = 'demo-01'
""").df()

demo_case_transactions
```

```python
con.execute("""
    SELECT
        case_ref,
        COUNT(*) AS transaction_count
    FROM case_transactions
    WHERE case_ref = 'demo-01'
    GROUP BY case_ref
""").df()
```
